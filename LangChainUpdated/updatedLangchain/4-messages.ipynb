```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.chat_models import init_chat_model

os.environ['GOOGLE_API_KEY'] = os.getenv("GOOGLE_API_KEY")

#os.environ['LLAMA_API_KEY'] = os.getenv('LLAMA_API_KEY');

#model=init_chat_model('google_genai:gemini-2.5-flash-lite')

model=ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')
```

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model(
    "groq:openai/gpt-oss-20b",
    api_key=os.getenv("LLAMA_API_KEY")
)

response = model.invoke("How are you?")
print(response.content)


```

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

# Use gpt-oss-120b or qwen/qwen3.8-27b on your Groq key
model = init_chat_model(
    "groq:openai/gpt-oss-120b",
    api_key=os.getenv("LLAMA_API_KEY")
)

print(model.invoke("Explain LangChain agents in two sentences.").content)
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages= [
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write me a poem on artificial intelligence"),
]

response = model.invoke(messages)
response.content
```

```python
system_message = SystemMessage("You are a coding expert")

messages= [
    system_message,
    HumanMessage("How do i create a REST API"),
]
response = model.invoke(messages)
response.content
```

```python
### Detailed info to the LLM through system message.  

from langchain.messages import SystemMessage, HumanMessage, AIMessage

system_message = SystemMessage("""You are a Senior Python developer with expertise in web frameworks. 
And also provide some code example and explain your reasoning. Be concise but thorough  in your explanations.  """)


messages= [
    system_message,
    HumanMessage("How do i create a REST API"),
]
response = model.invoke(messages)
print(response.content)




```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

# Create an AI message manually (e.g from conversation history)
ai_message = AIMessage("I'd be happy to help you with that question.")

messages= [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me ?"),
    ai_message,
     HumanMessage("Great, What's 2+2?"),
]
response = model.invoke(messages)
print(response.content)
```
